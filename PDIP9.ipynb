Importamos las librerías necesarias

Las librerías y el código son idénticos al del repositorio en github, únicamente se agrego la librería de urlib,request para poder consultar directo del url del github

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import urllib.request
```

Después vamos a descargar la imagen desde el repositorio de GitHub

```python
url = "https://raw.githubusercontent.com/dilp90/PDI/main/Imagenes/Im004_ACRIMA.jpg"
urllib.request.urlretrieve(url, "Im004_ACRIMA.jpg")
```

Después tenemos que descargar la imagen y convertir espacios de color, la imagen se carga por defecto en BGR, pero necesitamos pasarla al RGB para que Matplotlib la pueda mostrar de forma correcta; también necesitamos separar el canal de iluminación por lo que pasaremos la imagen al modelo HSV

```python
img_bgr = cv2.imread("Im004_ACRIMA.jpg")
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
```

Luego ocurre la mejora de calidad, se separan los tres canales del modelo HSV y se aplica la ecualización solamente en el canal Value para redistribuir el contraste y brillo.

Luego reconstruiremos la imagen

```python
h, s, v = cv2.split(img_hsv)
v_eq = cv2.equalizeHist(v)
img_hsv_eq = cv2.merge([h, s, v_eq])
img_eq_bgr = cv2.cvtColor(img_hsv_eq, cv2.COLOR_HSV2BGR)
img_eq_rgb = cv2.cvtColor(img_eq_bgr, cv2.COLOR_BGR2RGB)
```

Mostramos la imagen riginal y la ecualizada, junto con el histograma de intensidad del canal V antes y después.

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title('Imagen original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_eq_rgb)
plt.title('Imagen mejorada (Ecualización V)')
plt.axis('off')

plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 4))
plt.hist(v.ravel(), bins=256, range=[0, 256], color='thistle')
plt.title("Histograma del Canal V - Original")
plt.xlabel("Intensidad de píxel (0 a 255)")
plt.ylabel("Cantidad de píxeles")

plt.figure(figsize=(8, 4))
plt.hist(v_eq.ravel(), bins=256, range=[0, 256], color='mediumpurple')
plt.title("Histograma del Canal V - Ecualizado")
plt.xlabel("Intensidad de píxel (0 - 255)")
plt.ylabel("Cantidad de píxeles")

plt.tight_layout()
plt.show()
```

Luego ya se aplica el mismo proceminieto a dos imágenes médicas

```python
img_bgr = cv2.imread("Stomach-Crohn's Disease.jpg")
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
h, s, v = cv2.split(img_hsv)
v_eq = cv2.equalizeHist(v)
img_hsv_eq = cv2.merge([h, s, v_eq])
img_eq_bgr = cv2.cvtColor(img_hsv_eq, cv2.COLOR_HSV2BGR)
img_eq_rgb = cv2.cvtColor(img_eq_bgr, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title('Imagen original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_eq_rgb)
plt.title('Imagen mejorada (Ecualización V)')
plt.axis('off')

plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 4))
plt.hist(v.ravel(), bins=256, range=[0, 256], color='indianred')
plt.title("Histograma del Canal V - Original")
plt.xlabel("Intensidad de píxel (0 a 255)")
plt.ylabel("Cantidad de píxeles")

plt.figure(figsize=(8, 4))
plt.hist(v_eq.ravel(), bins=256, range=[0, 256], color='darkred')
plt.title("Histograma del Canal V - Ecualizado")
plt.xlabel("Intensidad de píxel (0 - 255)")
plt.ylabel("Cantidad de píxeles")

plt.tight_layout()
plt.show()
```
