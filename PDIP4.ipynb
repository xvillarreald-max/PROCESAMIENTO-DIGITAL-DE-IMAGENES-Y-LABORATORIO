```python
import cv2 as cv2
import numpy as np
import matplotlib.pyplot as plt
```

#Parte I.- Procesamiento geométrico
Ejercicio 1: Traslación

```python
# 1. Cargar una imagen médica en escala de grises (puede ser una radiografía o resonancia magnética).
image = cv2.imread("/content/RADIOGRAFÍA.webp", cv2.IMREAD_GRAYSCALE)
#IMAGEN ORIGINAL
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("ORIGINAL")
plt.axis("off")
```

```python
#2. Aplicar una traslación de (50, 30) píxeles en la dirección (x, y).
tx,ty = 50,30
M = np.float32([[1, 0, tx], [0, 1, ty]])
alto, ancho = image.shape[:2]

traslacion = cv2.warpAffine(image, M, (ancho, alto))
#IMAGEN TRASLADADA
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(traslacion, cmap="gray")
plt.title("TRASLACIÓN")
plt.axis("off")
```

```python
#3. Aplicar una traslación con valores decimales, por ejemplo, (20.5, 15.5) píxeles.
tx,ty = 200.50,120.30
M = np.float32([[1, 0, tx], [0, 1, ty]])
alto, ancho = image.shape[:2]

traslacion_decimal=cv2.warpAffine(image, M, (ancho, alto), flags=cv2.INTER_LINEAR)

#TRASLACION DECIMAL
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(traslacion_decimal, cmap="gray")
plt.title("TRASLACIÓN DECIMAL")
plt.axis("off")
```

Ejercicio 2: Rotación

```python
#1. Aplicar una rotación de 45° alrededor del centro de la imagen.
ang = 45
centro = (ancho // 2, alto // 2)
fun_rotacion = cv2.getRotationMatrix2D(centro, ang, 1)

Rotacion = cv2.warpAffine(image, fun_rotacion, (ancho, alto))
#COMPARACIÓN:
#IMAGEN ORGINAL
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("ORIGINAL")
plt.axis("off")

#IMAGEN ROTADA
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(Rotacion, cmap="gray")
plt.title("ROTACIÓN")
plt.axis("off")
```

Ejercicio 3: Escala

```python
#1. Aplicar un escalado al 150% de tamaño y otro al 50% de tamaño.
escala1 = 1.50 #150%
escala2 = 0.50 #50%
fun_escala1 = cv2.getRotationMatrix2D(centro, 0, escala1)
fun_escala2 = cv2.getRotationMatrix2D(centro, 0, escala2)

escala_1 = cv2.warpAffine(image, fun_escala1, (ancho, alto))
escala_2 = cv2.warpAffine(image, fun_escala2, (ancho, alto))

#COMPARACION
#IMAGEN ORIGINAL
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("ORIGINAL")
plt.axis("off")

#ESCALA 150
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(escala_1, cmap="gray")
plt.title("150%")
plt.axis("off")

#ESCALA 50%
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(escala_2, cmap="gray")
plt.title("50%")
plt.axis("off")
```

# Parte II.- Ecualización de histograma
Ejercicio 1: Ecualización

```python
# 1. Cargar una imagen médica con bajo contraste
imagen_bajocontraste = cv2.imread("/content/bajo.jpg", cv2.IMREAD_GRAYSCALE)
```

```python
# 2. Obtener y visualizar el histograma de la imagen original.
#HISTOGRAMA ORIGINAL
histograma = cv2.calcHist([imagen_bajocontraste], [0], None, [256], [0,256])

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(imagen_bajocontraste, cmap="gray")
plt.title("Original")
plt.axis("off")

plt.figure(figsize=(10,5))
plt.subplot(2,2,2)
plt.plot(histograma, color="black")
plt.title("Original")
plt.xlabel("Intensidad")
plt.ylabel("Frecuencia")


#ECUALIZACION
ecualizar = cv2.equalizeHist(imagen_bajocontraste)
histograma_ecualizado = cv2.calcHist([ecualizar], [0], None, [256], [0,256])

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(ecualizar, cmap="gray")
plt.title("Ecualizado")
plt.axis("off")

plt.figure(figsize=(10,5))
plt.subplot(2,2,2)
plt.plot(histograma_ecualizado, color="black")
plt.title("Ecualizado")
plt.xlabel("Intensidad")
plt.ylabel("Frecuencia")
```
